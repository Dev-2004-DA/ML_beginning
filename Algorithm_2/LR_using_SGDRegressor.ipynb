```python
# Createing a pipeline for Encoding and Scaling
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# splitting
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,[0,1,2,3,4]],df.iloc[:,5],test_size=0.2,random_state=90)

# ColumnTransformer
CT = ColumnTransformer(transformers=[('Encoding',OneHotEncoder(),[2]),
                                     ('Standarize',StandardScaler(),[1])],
                                     remainder='passthrough')

from sklearn.linear_model import SGDRegressor
sg = SGDRegressor(learning_rate='constant',eta0=0.001,max_iter=10000)

pipe = Pipeline(steps=[('trf',CT),
                       ('Model',sg)])
pipe.fit(X_train,y_train)

Y_pred = pipe.predict(X_test)
from sklearn.metrics import r2_score
print(r2_score(Y_pred,y_test))
```

```python
#  Using loop to check what happends when we increase eta0 or the learning rate 
for i in range(1,100):
    sg = SGDRegressor(learning_rate='constant',eta0=i*0.01,max_iter=100)
    pipe = Pipeline(steps=[('trf',CT),
                           ('Model',sg)])
    pipe.fit(X_train,y_train)
    Y_pred = pipe.predict(X_test)

    print(f"eta0={i*0.01},      {r2_score(Y_pred,y_test)}")

# we see as eta0 > 0.01  the R2 score become negative that is worst model
```

```python
import pandas as pd
import numpy as np 
```

```python
data = pd.read_csv(r"C:\Users\91972\Downloads\kc_house_data.csv")
```

```python
df = data.copy()
# df.info()
df.sample(2)
df['date'] = pd.to_datetime(df['date'])
df.sample()
df.info()
```

```python
# print(df.sample(50).to_string())

# df.drop('id',axis=1,inplace=True)

df2 = df.set_index(df['date'])      #Index = Date


df3 = df.drop('date',axis=1)
df3[['waterfront','view']]
df3=df3.drop(['waterfront','view'],axis=1)
df3.query('bedrooms == 3 and bathrooms == 1  and floors==1	and condition ==3 and ( yr_built < 1901 or yr_built > 1985 )')    # By query we found yr_built has influence on price
df3 = df3.drop(['zipcode','lat','long'],axis=1)
df3.columns
df3.drop('price',axis=1).corr()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df3.drop('price',axis=1),df3.loc[:,['price']],test_size=0.2,random_state=89)

# from sklearn.preprocessing import P
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler , PolynomialFeatures
from sklearn.linear_model import LinearRegression , SGDRegressor,Ridge,Lasso , ElasticNet
```

```python
sd = StandardScaler()
lr = LinearRegression()
```

```python
# Pipeline 1 for LR fitting

pipe1 = Pipeline(steps = [('Standardized',sd),
                          ('Model',lr)])

pipe1.fit(X_train,y_train)
y_trained1 = pipe1.predict(X_train)
y_pred1 = pipe1.predict(X_test)

from sklearn.metrics import r2_score
print(f"R2_Score on training data is {r2_score(y_train,y_trained1)*100}\nR2_Score on test data is {r2_score(y_test,y_pred1)*100}  ")
```

```python
# Pipeline 2 for LR fitting using Ridge
for i in range(5):
    pipe2 = Pipeline(steps = [('Standardized',sd),
                            ('Model',Ridge(alpha=i))])

    pipe2.fit(X_train,y_train)
    y_trained2 = pipe2.predict(X_train)
    y_pred2 = pipe2.predict(X_test)

    from sklearn.metrics import r2_score
    print(f"R2_Score on training data is {r2_score(y_train,y_trained2)*100}\nR2_Score on test data is {r2_score(y_test,y_pred2)*100}  ")
```

```python
# Pipeline 3 for LR fitting using Lasso
for i in range(100,500,50):
    pipe3 = Pipeline(steps = [('Standardized',sd),
                            ('Model',Lasso(alpha=i,max_iter=10000))])

    pipe3.fit(X_train,y_train)
    y_trained3 = pipe3.predict(X_train)
    y_pred3 = pipe3.predict(X_test)

    from sklearn.metrics import r2_score
    print(f"R2_Score on training data is {r2_score(y_train,y_trained3)*100}\nR2_Score on test data is {r2_score(y_test,y_pred3)*100}  ")
```

```python
# Pipeline 4 for LR fitting using ElasticNet

pipe4 = Pipeline(steps = [('Standardized',sd),
                        ('Model',ElasticNet(alpha=1,l1_ratio=0.95))])

pipe4.fit(X_train,y_train)
y_trained4 = pipe4.predict(X_train)
y_pred4 = pipe4.predict(X_test)

from sklearn.metrics import r2_score
print(f"R2_Score on training data is {r2_score(y_train,y_trained4)*100}\nR2_Score on test data is {r2_score(y_test,y_pred4)*100}  ")
```

```python
# Pipeline 5 for Polynomial Fitting
pipe5 = Pipeline(steps = [('Poly',PolynomialFeatures(degree=2)),
                          ('Standardized',sd),
                          ('Model',lr)])



pipe5.fit(X_train,y_train)
y_trained5 = pipe5.predict(X_train)
y_pred5 = pipe5.predict(X_test)

from sklearn.metrics import r2_score
print(f"R2_Score on training data is {r2_score(y_train,y_trained5)*100}\nR2_Score on test data is {r2_score(y_test,y_pred5)*100}  ")
```

```python
# Pipeline 6 for Polynomial Fitting using ElasticNet
pipe6 = Pipeline(steps = [('Poly',PolynomialFeatures(degree=2,include_bias=False)),
                          ('Standardized',sd),
                          ('Model',ElasticNet(alpha=1,l1_ratio=0.9,max_iter=10000))])



pipe6.fit(X_train,y_train)
y_trained6 = pipe6.predict(X_train)
y_pred6 = pipe6.predict(X_test)

from sklearn.metrics import r2_score
print(f"R2_Score on training data is {r2_score(y_train,y_trained6)*100}\nR2_Score on test data is {r2_score(y_test,y_pred6)*100}  ")
```

```python
from sklearn.decomposition import PCA
for j in (range(1,15)):
    pca = PCA(n_components=j)
    X_traina = X_train.copy()
    X_traineeed = pca.fit_transform(X_traina)
    print(f'{j}____{pca.explained_variance_ratio_.sum()}')

# only 7 PC is needed
```

```python
# Pipeline 6 for Polynomial Fitting using ElasticNet
from sklearn.preprocessing import PowerTransformer

pipe7 = Pipeline(steps = [
                          ('Standardized',sd),
                          ('PT',PowerTransformer()),
                          ('PCA',pca),
                          ('Poly',PolynomialFeatures(degree=2,include_bias=False)),
                          ('Model',LinearRegression())
                          ])



pipe7.fit(X_train,y_train)
y_trained7 = pipe7.predict(X_train)
y_pred7 = pipe7.predict(X_test)

from sklearn.metrics import r2_score
print(f"R2_Score on training data is {r2_score(y_train,y_trained7)*100}\nR2_Score on test data is {r2_score(y_test,y_pred7)*100}  ")
```

```python
y_trained = [y_trained1,y_trained2,y_trained3,y_trained4,y_trained5,y_trained6,y_trained7]
y_predicted = [y_pred1,y_pred2,y_pred3,y_pred4,y_pred5,y_pred6,y_pred7]

from  sklearn.metrics import mean_absolute_error , mean_squared_error

# for i in range(7):
    # print(f"BY MODEL {i+1}\nR2_Score on training data is {r2_score(y_train,y_trained[i])*100}\nR2_Score on test data is {r2_score(y_test,y_predicted[i])*100}\n ")
R2_train = [0,1,2,3,4,5,6]
R2_test = [0,1,2,3,4,5,6]
mae_train = [0,1,2,3,4,5,6]
mae_test = [0,1,2,3,4,5,6]
rmse_train = [0,1,2,3,4,5,6]
rmse_test = [0,1,2,3,4,5,6]
for i in range(7):
    R2_train[i] = r2_score(y_train,y_trained[i])*100
    R2_test[i] = r2_score(y_test,y_predicted[i])*100
    mae_train[i] = mean_absolute_error(y_train,y_trained[i])*100
    mae_test[i] = mean_absolute_error(y_test,y_predicted[i])*100
    rmse_train[i] = np.sqrt(mean_squared_error(y_train,y_trained[i]))*100
    rmse_test[i] = np.sqrt(mean_squared_error(y_test,y_predicted[i]))*100

Model_test = pd.DataFrame({
                          'Model' : ['LR','LR + Ridge','LR + Lasso','LR + ElasticNet','Polynomial + LR','Polynomial + ElasticNet' , 'Polynomial + PCA + LR'],
                          ('Training','R2'):np.round(R2_train,2),
                          ('Training','MAE'):np.round(mae_train,2),
                          ('Training','RMSE'):np.round(rmse_train,2),
                          ('Test','R2'):np.round(R2_test,2),
                          ('Test','MAE'):np.round(mae_test,2),
                          ('Test','RMSe'):np.round(rmse_test,2)
                          })
Model_test
```

```python
print(f"On observing the dataframe Model_test we found that the most suitable model among these is:\n Polynomial + PCA + LR with corresponding accuracy score\n{Model_test.iloc[6,:]}")
```

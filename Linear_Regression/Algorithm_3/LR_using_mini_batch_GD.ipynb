## Here We will fit Linear Model using Mini Batch Gradient Regressor 
1. 1st we have to check the data
2. Then have to do Univariate and Bivarite Analysis ( also done in Linear Regression)
3. Then the very important thing have to apply Feature Scaling 
4. Then set a loop for epoch(for 20 epochs)
5. Then again an internal loop to divide dataset into mini batches for each epoch
6. Lastly apply Algorithm using Partial_fit() to every chunk

```python
# Createing a pipeline for Encoding and Scaling
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# splitting
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,[0,1,2,3,4]],df.iloc[:,5],test_size=0.2,random_state=90)

# ColumnTransformer
CT = ColumnTransformer(transformers=[('Encoding',OneHotEncoder(),[2]),
                                     ('Standarize',StandardScaler(),[1])],
                                     remainder='passthrough')
X_scaled = CT.fit_transform(X_train)
X_scaled = pd.DataFrame(X_scaled)
```

```python
from sklearn.linear_model import SGDRegressor
mini_batch_gd = SGDRegressor(max_iter=1,eta0=0.01,learning_rate='constant')

batch_size=3000
epochs = 200

# Loop for 20 epochs
for epochs in range(epochs):
    indices = np.random.permutation(range(len(X_train)))                    # Shuffling list of index
    X_shuffled = X_scaled.iloc[indices,:].reset_index(drop=True)
    y_shuffled = y_train.iloc[indices].reset_index(drop=True)

    # Loop for mini batches in each epoch
    for i in range(0,len(X_train),batch_size):
        X_mini = X_shuffled.iloc[i:i+batch_size,:]
        y_mini = y_shuffled.iloc[i:i+batch_size]

        # Partial fiting the model for each mini batch
        mini_batch_gd.partial_fit(X_mini,y_mini)
```
